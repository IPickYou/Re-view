```python
from transformers import T5ForConditionalGeneration, PreTrainedTokenizerFast

# 모델 로드
model_id = "aimer3152/summary_model"
tokenizer = PreTrainedTokenizerFast.from_pretrained(model_id)
model = T5ForConditionalGeneration.from_pretrained(model_id)

# 요약할 텍스트
text = """
어 소통 능력은 지금까지 경험한 바로는 음 아랫사람이던 윗사람이던 개별적으로 많은 대화를 통해서 그 사람을 알아가고 경험치를 이해하고 그리고 서로 대화를 통해서 이제 전달하고 주고받으면서 그 능력이 키워진다고 생각합니다. 그래서 쫌 더 많은 시간을 틈틈이 할애해서 서로 서로 자주 대화할 수 있는 그러한 시간들을 많이 가지는 게 소통 능력을 키우는 데 효율적이라고 생각합니다. 그리고 외부적으로도 다른 카페를 가입을 통해서 뭐 독서 모임을 하다든지 다른 외부적인 취미 활동을 통해서 제 분야와 다른 사람들과 소통을 하고 는 그 경험을 통해서 좀 더 개발할 수 있다고 생각이 듭니다.

"""

# T5 스타일의 입력 프롬프트 구성
input_text = "summarize: " + text.strip()

# 토크나이징
inputs = tokenizer.encode(input_text, return_tensors="pt", max_length=512, truncation=True)

# 요약 생성
summary_ids = model.generate(inputs, max_length=100, min_length=10, length_penalty=2.0, num_beams=4, early_stopping=True)

# 결과 디코딩
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

# 출력
print("요약 결과:", summary)
```
