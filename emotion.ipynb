```python
from ultralytics import YOLO

model = YOLO("yolo12n.pt")

model.train(data = 'c:/big_data/ipickyou/review/dataset/emotion.yaml',
            epochs = 150,
            patience = 20,
            project = 'emotion_od',
            name = 'emotion_model',
            device = 0)
```

```python
from ultralytics import YOLO
import cv2

model = YOLO("emotion_od/emotion_model/weights/best.pt")

results_generator = model.predict(
    source = 0,
    show = True,
    conf = 0.3,
    stream = True
)

for result in results_generator:

    print("Boxes :", result.boxes.xyxy)
    print("Classes :", result.boxes.cls)
    print("Confidences :", result.boxes.conf)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cv2.destroyAllWindows()
```
